```python
# 라이브러리 불러오기
import pandas as pd

TITANIC_URL = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
```

```python
# 결측치
titanic.isnull().sum() 
# 컬럼별 결측치(조건이 True)의 합을 확인.
```

```python
# 결측치 재우기
df = titanic.copy() # 사본을 만듬
df['Age'] = df['Age'].fillna(df['Age'].median()) # 비어있는 값은 중앙값으로 채움.
```

```python
df.isnull().sum()
# df['Age'] = df['Age'].dropna() # 비어있는 값 다 날러버려 => 177개의 행이 날라간다. 
```

```python
## groupby 그룹별로 보기
df.groupby('Sex')['Survived'].mean() # 성별 기준으로 그룹화
# 그룹화 한 함수의 결과(Object)에 집계를 함.
# 생존자의 평균
```

```python
# 집계 함수
df.groupby('Pclass').agg(생존율=('Survived', 'mean'), 인원=('Survived', 'count'))
```

```python
# 남녀 성비를 비율로
df['Sex'].value_counts(normalize=True)
```

```python
# 숫자로 카운팅
df['Sex'].value_counts()
```
